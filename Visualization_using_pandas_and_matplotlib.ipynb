```python
plt.figure(figsize=(12,6))
for column in df.columns[1:]:
    plt.plot(df['Year'],df[column], label=column)

plt.title('Gas Prices Over Time(1990-2008) ')
plt.xlabel('Year')
plt.ylabel('Gas Price')
plt.legend()
plt.grid(True)
plt.show()
```

```python
avg_prices = df.iloc[:,1:].mean()

plt.figure(figsize=(10,5))

avg_prices.plot(kind='bar')

plt.title('Average Gas Price by Country')
plt.xlabel('Country')
plt.ylabel('Average Price')
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['USA'], bins=8)

plt.title('Distribution of USA Gas Prices')
plt.xlabel('Price')
plt.ylabel('Frequency')

plt.show()
```

```python
plt.figure(figsize=(12,6))

df.iloc[:,1:].boxplot()

plt.title('Gas Price Distribution by Country')
plt.ylabel('Price')

plt.show()
```

```python
prices_2008 = df[df['Year'] == 2008].iloc[:,1:].T

prices_2008.plot(kind='bar', legend=False)

plt.title('Gas Prices in 2008')
plt.xlabel('Country')
plt.ylabel('Price')

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(df['Year'], df['USA'], marker='o', label='USA')
plt.plot(df['Year'], df['Canada'], marker='o', label='Canada')

plt.title("USA vs Canada Gas Prices")
plt.xlabel("Year")
plt.ylabel("Price")
plt.legend()
plt.grid(True)

plt.show()
```

```python
corr = df.iloc[:,1:].corr()

plt.figure(figsize=(10,8))

plt.imshow(corr,cmap="PuRd")

plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title("Correlation Between Countries")

plt.show()
```

```python
plt.figure(figsize=(10,7))

avg_price.sort_values().plot(kind='barh')

plt.title('Country Ranking by Average Gas Price')
plt.xlabel('Average Gas Price')
plt.ylabel('Country')

plt.grid(axis='x')
plt.show()
```

```python
plt.figure(figsize=(14,7))

selected_countries = ['USA', 'Canada', 'UK', 'Germany']

plt.stackplot(
    df['Year'],
    df['USA'],
    df['Canada'],
    df['UK'],
    df['Germany'],
    labels=selected_countries
)

plt.title('Gas Price Trends of Selected Countries')
plt.xlabel('Year')
plt.ylabel('Gas Price')
plt.legend(loc='upper left')

plt.show()
```

```python
plt.figure(figsize=(9,6))

plt.scatter(df['USA'], df['Canada'])

plt.title('Relationship Between USA and Canada Gas Prices')
plt.xlabel('USA Gas Price')
plt.ylabel('Canada Gas Price')

plt.grid(True)
plt.show()
```

```python
price_2008 = df[df['Year'] == 2008].iloc[0, 1:]

plt.figure(figsize=(9,9))

plt.pie(
    price_2008,
    labels=price_2008.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Gas Price Distribution by Country in 2008')

plt.show()
```
